```python
numbers = [1, 3, 2, 6]
window_size = 2

# 1. Start with the sum of the first window [1, 3]
window_sum = numbers[0] + numbers[1]  # 1 + 3 = 4
print("First window sum:", window_sum)

# 2. Slide the window right one by one!
for i in range(len(numbers) - window_size):
    # Subtract the element leaving, add the element joining!
    window_sum = window_sum - numbers[i] + numbers[i + window_size]
    print("Next window sum:", window_sum)
```
