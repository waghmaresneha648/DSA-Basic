```python
from collections import deque
line = deque()
line.append("Alice ")
line.append("Bob ")
line.append("Charlie ")

print("Line right now:", line)
# 3. Serve the first person in line (DEQUEUE)
served = line.popleft()  # Takes from the FRONT!
print("Served:", served)
print("Line remaining:", line)
```
